```python
# %% [markdown]
# # 🧪 Meme Model Performance Evaluation (Pre-tokenized Data Ver.)
# 
# 이 노트북은 로컬에 저장된 KcELECTRA 기반 밈 분류 모델(`src/talaria/assets/model`)의 성능을 정량적으로 평가합니다.
# 데이터가 이미 Tokenize된 상태(`val.pkl`)이므로, 전처리 과정을 건너뛰고 평가합니다.

# %%
import os
import sys
import torch
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from tqdm import tqdm
from sklearn.metrics import classification_report, confusion_matrix
from transformers import AutoTokenizer, ElectraPreTrainedModel, ElectraModel, AutoConfig

# 경고 메시지 억제
import warnings
warnings.filterwarnings("ignore")

# %% [markdown]
# ## 1. 환경 설정 및 데이터 로드

# %%
# 1. 경로 설정
# 현재 위치가 talaria-ai/test 라고 가정하고 상위 경로 설정
# 만약 경로 에러가 나면 절대 경로로 수정하세요.
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
MODEL_PATH = os.path.join(PROJECT_ROOT, "src/talaria/assets/model")
VAL_DATA_PATH = os.path.join(PROJECT_ROOT, "model/train_data/val.pkl")

print(f"📂 Project Root: {PROJECT_ROOT}")
print(f"📂 Model Path: {MODEL_PATH}")
print(f"📂 Validation Data: {VAL_DATA_PATH}")

# 2. Device 설정
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"⚡ Device: {device}")

# %% [markdown]
# ## 2. 모델 클래스 정의

# %%
class MemeMultiTaskModel(ElectraPreTrainedModel):
    def __init__(self, config):
        super().__init__(config)
        self.num_labels_cls = 4 
        self.num_labels_ner = 3
        self.electra = ElectraModel(config)
        self.dropout = torch.nn.Dropout(config.hidden_dropout_prob)
        self.classifier = torch.nn.Linear(config.hidden_size, self.num_labels_cls)
        self.ner_classifier = torch.nn.Linear(config.hidden_size, self.num_labels_ner)
        
    def forward(self, input_ids=None, attention_mask=None, token_type_ids=None):
        discriminator_hidden_states = self.electra(input_ids, attention_mask, token_type_ids)
        sequence_output = discriminator_hidden_states[0]
        pooled_output = sequence_output[:, 0, :] 
        pooled_output = self.dropout(pooled_output)
        logits_cls = self.classifier(pooled_output)
        logits_ner = self.ner_classifier(sequence_output)
        return logits_cls, logits_ner

# %% [markdown]
# ## 3. 모델 및 데이터 로드

# %%
# 모델 & 토크나이저 로드
try:
    config = AutoConfig.from_pretrained(MODEL_PATH)
    tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
    model = MemeMultiTaskModel.from_pretrained(MODEL_PATH, config=config)
    model.to(device)
    model.eval()
    print("✅ 모델 로드 성공!")
except Exception as e:
    print(f"❌ 모델 로드 실패: {e}")
    # 모델 경로가 틀렸을 경우를 대비해 예외 처리
    raise e

# 데이터 로드
if os.path.exists(VAL_DATA_PATH):
    val_data = pd.read_pickle(VAL_DATA_PATH)
    print(f"✅ 검증 데이터 로드 완료: {len(val_data)} rows")
    print(f"👉 컬럼 확인: {val_data.columns.tolist()}")
else:
    print(f"❌ 데이터 파일 없음: {VAL_DATA_PATH}")

# Label Mapping
id2label = {0: "SARCASM", 1: "ADMIRATION", 2: "SHOCK", 3: "NEUTRAL"}
id2ner = {0: "O", 1: "B-MEME", 2: "I-MEME"}

# %% [markdown]
# ## 4. 추론 실행 (이미 토큰화된 데이터 사용)

# %%
BATCH_SIZE = 32

true_labels = []
pred_labels = []

true_ner_list = []
pred_ner_list = []

# 원문 복원용 (Error Analysis를 위해)
decoded_texts_sample = [] 
sample_limit = 5 # 5개만 샘플로 저장

print("🚀 추론 시작...")

with torch.no_grad():
    for i in tqdm(range(0, len(val_data), BATCH_SIZE)):
        batch = val_data.iloc[i:i+BATCH_SIZE]
        
        # [수정됨] 이미 숫자(Tensor)로 되어 있으므로 tokenizer() 호출 불필요
        # 리스트 형태의 데이터를 텐서로 변환
        input_ids = torch.tensor(batch['input_ids'].tolist()).to(device)
        attention_mask = torch.tensor(batch['attention_mask'].tolist()).to(device)
        token_type_ids = torch.tensor(batch['token_type_ids'].tolist()).to(device)
        labels = batch['labels'].tolist()
        ner_labels = batch['ner_labels'].tolist()
        
        # 모델 입력
        logits_cls, logits_ner = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        
        # 1. 감정 분류 예측 (Class Prediction)
        batch_preds = torch.argmax(logits_cls, dim=1).cpu().numpy()
        pred_labels.extend(batch_preds)
        true_labels.extend(labels)
        
        # 2. NER 예측
        batch_ner_preds = torch.argmax(logits_ner, dim=2).cpu().numpy()
        
        # NER 평가 시 -100(Padding/Special Token) 제외하고 비교
        for idx, (pred_seq, true_seq) in enumerate(zip(batch_ner_preds, ner_labels)):
            valid_preds = []
            valid_trues = []
            for p, t in zip(pred_seq, true_seq):
                if t != -100: # -100이 아닌 유효한 라벨만 평가
                    valid_preds.append(p)
                    valid_trues.append(t)
            
            true_ner_list.extend(valid_trues)
            pred_ner_list.extend(valid_preds)

            # 샘플 텍스트 복원 (처음 몇 개만)
            if len(decoded_texts_sample) < sample_limit:
                decoded_text = tokenizer.decode(input_ids[idx], skip_special_tokens=True)
                decoded_texts_sample.append({
                    "text": decoded_text,
                    "true": id2label[labels[idx]],
                    "pred": id2label[batch_preds[idx]]
                })

print("🏁 추론 완료")

# %% [markdown]
# ## 5. 📊 데이터 불균형 분석 (Data Imbalance)

# %%
# 실제 라벨 분포 확인
plt.figure(figsize=(10, 5))
label_names = [id2label[l] for l in val_data['labels']]
# 순서 정렬을 위해
order = ["SARCASM", "ADMIRATION", "SHOCK", "NEUTRAL"]
sns.countplot(x=label_names, order=order, palette='viridis')

plt.title('Validation Data Class Distribution')
plt.xlabel('Emotion Label')
plt.ylabel('Count')
plt.show()

# 수치 출력
print("=== 클래스별 데이터 개수 ===")
counts = val_data['labels'].value_counts().sort_index()
for idx, count in counts.items():
    print(f"{id2label[idx]}: {count}개 ({count/len(val_data)*100:.1f}%)")

# %% [markdown]
# ## 6. 📉 감정 분류 상세 평가 (Confusion Matrix)

# %%
print("\n" + "="*50)
print("📑 [성적표 1] 감정 분류 상세 리포트")
print("="*50)
print(classification_report(true_labels, pred_labels, target_names=[id2label[i] for i in range(4)], digits=4))

# Confusion Matrix
cm = confusion_matrix(true_labels, pred_labels)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=[id2label[i] for i in range(4)], 
            yticklabels=[id2label[i] for i in range(4)])
plt.title('Emotion Classification Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()

# %% [markdown]
# ## 7. 📝 오답 노트 (Error Analysis)
# 실제로 어떤 문장을 틀렸는지 확인합니다. (위에서 복원한 텍스트 활용)

# %%
print("=== 예측 샘플 확인 (텍스트 복원) ===")
for sample in decoded_texts_sample:
    status = "✅" if sample['true'] == sample['pred'] else "❌"
    print(f"{status} [실제: {sample['true']} vs 예측: {sample['pred']}] 문장: {sample['text']}")

# %% [markdown]
# ## 8. 🎯 밈 추출(NER) 성능 평가

# %%
print("\n" + "="*50)
print("📑 [성적표 2] 밈 단어 추출 성능 (Token Level)")
print("="*50)

# 0, 1, 2 에 해당하는 태그 이름
target_names = [id2ner[i] for i in sorted(list(set(true_ner_list)))]

print(classification_report(true_ner_list, pred_ner_list, target_names=target_names, digits=4))
```

```python
import os
import sys
import pandas as pd
from transformers import AutoTokenizer

# ==========================================
# 1. 경로 설정
# ==========================================
current_path = os.getcwd()
if "talaria-ai" in current_path:
    project_root = current_path
    while os.path.basename(project_root) != "talaria-ai":
        project_root = os.path.dirname(project_root)
        if len(project_root) < 3: break
    PROJECT_ROOT = project_root
else:
    PROJECT_ROOT = os.path.abspath(os.path.join(current_path, ".."))

MODEL_PATH = os.path.join(PROJECT_ROOT, "src/talaria/assets/model")
VAL_DATA_PATH = os.path.join(PROJECT_ROOT, "model/train_data/val.pkl")

# ==========================================
# 2. RoBERTa 바이트 디코더 (필살기)
# ==========================================
def get_byte_decoder():
    # GPT-2/RoBERTa의 바이트->유니코드 매핑 테이블 생성 (Standard Logic)
    bs = list(range(ord("!"), ord("~")+1)) + list(range(ord("¡"), ord("¬")+1)) + list(range(ord("®"), ord("ÿ")+1))
    cs = bs[:]
    n = 0
    for b in range(2**8):
        if b not in bs:
            bs.append(b)
            cs.append(2**8 + n)
            n += 1
    cs = [chr(n) for n in cs]
    return dict(zip(cs, bs)) # Key: 외계어(Ġ 등), Value: 실제 바이트 숫자

BYTE_DECODER = get_byte_decoder()

def decode_roberta(text):
    # 1. 토크나이저가 디코딩하며 넣은 '가짜 공백' 제거
    # (RoBERTa는 공백을 'Ġ'로 표현하므로, 텍스트 상의 ' '는 토크나이저가 토큰 사이에 끼워넣은 것임)
    text_no_space = text.replace(" ", "")
    
    # 2. 바이트 역변환
    try:
        bytes_list = []
        for c in text_no_space:
            if c in BYTE_DECODER:
                bytes_list.append(BYTE_DECODER[c])
            else:
                # 맵핑에 없는 문자는 그대로 둠 (혹은 무시)
                # 보통 한글 깨짐은 여기서 다 걸림
                pass
        
        # 3. UTF-8로 최종 디코딩
        return bytes(bytes_list).decode("utf-8", errors='replace')
    except Exception as e:
        return f"[Decode Error] {e} / Original: {text}"

# ==========================================
# 3. 검증 실행
# ==========================================
def verify_text_decoding():
    if not os.path.exists(VAL_DATA_PATH):
        print(f"❌ 오류: 파일 없음 - {VAL_DATA_PATH}")
        return
    
    df = pd.read_pickle(VAL_DATA_PATH)
    print(f"✅ 데이터 로드 완료 ({len(df)}개)")

    try:
        tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
        print(f"✅ 토크나이저 로드: {type(tokenizer).__name__}")
    except:
        print("❌ 토크나이저 로드 실패")
        return

    print("\n" + "="*60)
    print("🕵️‍♂️ 텍스트 복원 테스트 (Top 5)")
    print("="*60)

    for i in range(5):
        row = df.iloc[i]
        input_ids = row['input_ids']
        
        # 1. Raw Decoding (외계어)
        raw_text = tokenizer.decode(input_ids, skip_special_tokens=True)
        
        # 2. Fix Decoding (한글)
        fixed_text = decode_roberta(raw_text)
        
        print(f"[Sample {i+1}]")
        print(f"👉 Raw: {raw_text[:50]}...") # 너무 길면 자름
        print(f"👉 Fix: {fixed_text}")
        
        if "ì" in fixed_text: # 여전히 깨져있으면
            print("   ❌ (복구 실패)")
        else:
            print("   ✨ (복구 성공!)")
        print("-" * 60)

if __name__ == "__main__":
    verify_text_decoding()
```

```python
import os
import sys
import time
import json
import pandas as pd
import torch
import numpy as np
from tqdm import tqdm
from openai import OpenAI
from transformers import AutoTokenizer, AutoConfig, ElectraPreTrainedModel, ElectraModel

# ==========================================
# 1. 설정 (Configuration)
# ==========================================
current_path = os.getcwd()
if "talaria-ai" in current_path:
    project_root = current_path
    while os.path.basename(project_root) != "talaria-ai":
        project_root = os.path.dirname(project_root)
        if len(project_root) < 3: break
    PROJECT_ROOT = project_root
else:
    PROJECT_ROOT = os.path.abspath(os.path.join(current_path, ".."))

MODEL_PATH = os.path.join(PROJECT_ROOT, "src/talaria/assets/model")
VAL_DATA_PATH = os.path.join(PROJECT_ROOT, "model/train_data/val.pkl")
RESULT_FILE = "../test/final_model_comparison_results.csv"

client = OpenAI()
MODELS_TO_TEST = ["gpt-4o-mini", "gpt-4o", "gpt-5.1"] 
ID2LABEL = {0: "SARCASM", 1: "ADMIRATION", 2: "SHOCK", 3: "NEUTRAL"}

# ==========================================
# 2. 유틸리티 (텍스트 복원 & 모델)
# ==========================================
def get_byte_decoder():
    bs = list(range(ord("!"), ord("~")+1)) + list(range(ord("¡"), ord("¬")+1)) + list(range(ord("®"), ord("ÿ")+1))
    cs = bs[:]
    n = 0
    for b in range(2**8):
        if b not in bs:
            bs.append(b)
            cs.append(2**8 + n)
            n += 1
    cs = [chr(n) for n in cs]
    return dict(zip(cs, bs))

BYTE_DECODER = get_byte_decoder()

def decode_roberta(text):
    text_no_space = text.replace(" ", "")
    try:
        bytes_list = []
        for c in text_no_space:
            if c in BYTE_DECODER:
                bytes_list.append(BYTE_DECODER[c])
        return bytes(bytes_list).decode("utf-8", errors='replace')
    except:
        return text

class MemeMultiTaskModel(ElectraPreTrainedModel):
    def __init__(self, config):
        super().__init__(config)
        self.num_labels_cls = 4 
        self.num_labels_ner = 3
        self.electra = ElectraModel(config)
        self.dropout = torch.nn.Dropout(config.hidden_dropout_prob)
        self.classifier = torch.nn.Linear(config.hidden_size, self.num_labels_cls)
        self.ner_classifier = torch.nn.Linear(config.hidden_size, self.num_labels_ner)
        
    def forward(self, input_ids=None, attention_mask=None, token_type_ids=None):
        discriminator_hidden_states = self.electra(input_ids, attention_mask, token_type_ids)
        sequence_output = discriminator_hidden_states[0]
        pooled_output = sequence_output[:, 0, :] 
        pooled_output = self.dropout(pooled_output)
        logits_cls = self.classifier(pooled_output)
        logits_ner = self.ner_classifier(sequence_output)
        return logits_cls, logits_ner

# ==========================================
# 3. 데이터 준비
# ==========================================
def get_test_samples(limit=None):
    if not os.path.exists(VAL_DATA_PATH):
        print("❌ 데이터 파일이 없습니다.")
        return []
        
    df = pd.read_pickle(VAL_DATA_PATH)
    tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
    
    samples = []
    df_sample = df.sample(frac=1, random_state=42).reset_index(drop=True)
    if limit and str(limit).lower() != 'all':
        df_sample = df_sample.head(int(limit))
    
    print(f">>> 데이터 로딩 및 복원 중 ({len(df_sample)}개)...")
    
    for idx, row in tqdm(df_sample.iterrows(), total=len(df_sample)):
        input_ids = row['input_ids']
        raw_decoded = tokenizer.decode(input_ids, skip_special_tokens=True)
        text = decode_roberta(raw_decoded)
        
        meme_tokens = []
        for t_id, n_label in zip(input_ids, row['ner_labels']):
            if n_label in [1, 2]:
                meme_tokens.append(t_id)
        raw_meme = tokenizer.decode(meme_tokens, skip_special_tokens=True)
        true_meme = decode_roberta(raw_meme).strip()
        true_emotion = ID2LABEL[row['labels']]
        
        samples.append({
            "id": idx, "text": text, "true_emotion": true_emotion, "true_meme": true_meme, "raw_row": row
        })
    return samples

# ==========================================
# 4. 추론 함수
# ==========================================
def run_local_model(model, sample, tokenizer, device):
    row = sample['raw_row']
    start_time = time.time()
    
    input_ids = torch.tensor([row['input_ids']]).to(device)
    attention_mask = torch.tensor([row['attention_mask']]).to(device)
    token_type_ids = torch.tensor([row['token_type_ids']]).to(device)
    
    with torch.no_grad():
        logits_cls, logits_ner = model(input_ids, attention_mask, token_type_ids)
        pred_cls = torch.argmax(logits_cls, dim=1).item()
        pred_emotion = ID2LABEL[pred_cls]
        
        pred_ner = torch.argmax(logits_ner, dim=2)[0].cpu().numpy()
        meme_tokens = []
        for t_id, n_label in zip(row['input_ids'], pred_ner):
            if n_label in [1, 2]:
                meme_tokens.append(t_id)
        raw_meme_pred = tokenizer.decode(meme_tokens, skip_special_tokens=True)
        pred_meme = decode_roberta(raw_meme_pred).strip()
        
    end_time = time.time()
    return pred_emotion, pred_meme, (end_time - start_time) * 1000

def run_openai_model_safe(model_name, text):
    start_time = time.time()
    prompt = f"""
    Analyze the Korean sentence.
    Sentence: "{text}"
    
    Task 1: Extract the 'Meme Word' (slang).
    Task 2: Classify emotion: [SARCASM, ADMIRATION, SHOCK, NEUTRAL].
    
    Output JSON: {{"meme": "string", "emotion": "string"}}
    """
    
    max_retries = 3
    for attempt in range(max_retries):
        try:
            response = client.chat.completions.create(
                model=model_name,
                messages=[{"role": "user", "content": prompt}],
                temperature=0,
                response_format={"type": "json_object"}
            )
            res = json.loads(response.choices[0].message.content)
            end_time = time.time()
            return res.get("emotion", "ERR"), res.get("meme", ""), (end_time - start_time) * 1000
        except Exception as e:
            if "rate limit" in str(e).lower() or "429" in str(e):
                time.sleep(2 * (attempt + 1))
                continue
            return "ERROR", "", 0
    return "ERROR_TIMEOUT", "", 0

# ==========================================
# 5. 결과 저장 및 분석 (핵심 수정됨)
# ==========================================
def save_result(sample, model_name, emo, meme, time_ms):
    # 1. 감정 정확도
    emo_correct = (emo.upper() == sample['true_emotion'])
    
    # 2. 밈 정확도 (양방향 포함 관계 확인 - Soft Match)
    t_meme = sample['true_meme'].strip()
    p_meme = meme.strip()
    
    if not t_meme: # 정답 밈이 없는 경우
        meme_correct = not p_meme # 예측도 없어야 정답
    else:
        # [핵심] 정답이 예측에 있거나, 예측이 정답에 있으면 인정
        # 예: "자강두천이" in "자강두천" (X) -> "자강두천" in "자강두천이" (O) => Pass
        meme_correct = (t_meme in p_meme) or (p_meme in t_meme)
    
    new_row = pd.DataFrame([{
        "id": sample['id'],
        "text": sample['text'],
        "model": model_name,
        "emotion": emo,
        "meme": meme,
        "time_ms": time_ms,
        "emotion_correct": emo_correct,
        "meme_correct": meme_correct
    }])
    
    # 헤더 처리를 위해 파일 존재 여부 확인
    header = not os.path.exists(RESULT_FILE)
    new_row.to_csv(RESULT_FILE, mode='a', header=header, index=False)

def analyze_results():
    if not os.path.exists(RESULT_FILE): return
    df = pd.read_csv(RESULT_FILE)
    
    # 에러 난 건 통계에서 제외할 수도 있음
    df = df[df['emotion'] != "ERROR"]
    
    summary = df.groupby("model").agg(
        Emo_Accuracy=("emotion_correct", "mean"),
        Meme_Accuracy=("meme_correct", "mean"),
        Avg_Latency_ms=("time_ms", "mean"),
        Count=("id", "count")
    ).sort_values("Avg_Latency_ms")
    
    print("\n🏆 최종 성적표 (양방향 포함 기준 적용) 🏆")
    print(summary)

# ==========================================
# 6. 메인 실행
# ==========================================
def main():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f">>> 로컬 모델 로딩 중... ({device})")
    
    try:
        config = AutoConfig.from_pretrained(MODEL_PATH)
        local_model = MemeMultiTaskModel.from_pretrained(MODEL_PATH, config=config).to(device)
        local_model.eval()
        local_tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
    except Exception as e:
        print(f"❌ 모델 로드 에러: {e}")
        return

    # 기존 결과 파일 삭제 (새로 시작 시)
    if os.path.exists(RESULT_FILE):
        print(f"⚠️ 기존 결과 파일({RESULT_FILE})이 있습니다. 이어쓰기 됩니다.")

    # 전체 실행
    user_input = input(">>> 테스트할 데이터 개수를 입력하세요 (숫자 or 'all'): ").strip()
    final_samples = get_test_samples(limit=user_input)
    
    print(f"\n🚀 대규모 테스트 시작 ({len(final_samples)}개)")
    print("-" * 60)

    for sample in tqdm(final_samples, desc="Processing"):
        # 1. Local
        l_emo, l_meme, l_time = run_local_model(local_model, sample, local_tokenizer, device)
        save_result(sample, "Talaria-Local", l_emo, l_meme, l_time)

        # 2. OpenAI
        for model_name in MODELS_TO_TEST:
            o_emo, o_meme, o_time = run_openai_model_safe(model_name, sample['text'])
            save_result(sample, model_name, o_emo, o_meme, o_time)
            time.sleep(0.5) # Rate Limit 방지용 텀

    print("\n✅ 테스트 완료!")
    analyze_results()

if __name__ == "__main__":
    main()
```

```python
import os
import pandas as pd
import json
import random
from openai import OpenAI
from tqdm import tqdm

# ==========================================
# 1. 설정 (Configuration)
# ==========================================
# OpenAI API 키가 환경변수에 없다면 아래에 직접 입력하세요.
# os.environ["OPENAI_API_KEY"] = "sk-..."
client = OpenAI()

OUTPUT_FILE = "../test/why_papago_final_report.csv"

# ==========================================
# 2. 데이터 입력 (여기를 채워주세요!)
# ==========================================
# 웹 번역기(Papago, DeepL, Google)를 켜두고 결과를 복사/붙여넣기 하세요.

# [SET A] 한국어 -> 영어 번역 테스트


# ==========================================
# 3. LLM 심판 (GPT-4o Evaluation Logic)
# ==========================================
def evaluate_translations(dataset, source_lang="Korean"):
    results = []
    
    target_lang = "English" if source_lang == "Korean" else "Korean"
    print(f"\n🚀 [Real Blind Test] 번역 품질 평가 시작 ({source_lang} -> {target_lang})")
    
    for item in tqdm(dataset, desc=f"Evaluating {source_lang}"):
        if not item["papago"] or not item["deepl"] or not item["google"]:
            continue

        # 1. 후보군 생성 및 셔플 (이름 숨기기)
        candidates = [
            {"name": "Papago", "text": item["papago"]},
            {"name": "DeepL", "text": item["deepl"]},
            {"name": "Google", "text": item["google"]}
        ]
        random.shuffle(candidates) # 순서를 무작위로 섞음
        
        # 2. 익명화된 프롬프트 생성
        prompt = f"""
        You are a professional translator and linguistic judge specializing in {source_lang} and {target_lang}.
        Compare the following translations for the source sentence.
        
        [Context]
        Category: {item['category']}
        Source ({source_lang}): "{item['source']}"

        [Anonymous Translations]
        A: "{candidates[0]['text']}"
        B: "{candidates[1]['text']}"
        C: "{candidates[2]['text']}"

        [Task]
        1. Evaluate each translation (A, B, C) on a scale of 1-10.
           - Focus on: Naturalness, Nuance (slang/idioms), and Business Tone.
        2. Select the WINNER (A, B, or C).
        3. Explain WHY briefly.

        Output JSON format:
        {{
            "scores": {{"A": int, "B": int, "C": int}},
            "winner": "A" or "B" or "C",
            "reason": "string"
        }}
        """
        
        try:
            response = client.chat.completions.create(
                model="gpt-4o",
                messages=[{"role": "user", "content": prompt}],
                response_format={"type": "json_object"},
                temperature=0
            )
            res = json.loads(response.choices[0].message.content)
            
            # 3. 결과 복호화 (A, B, C가 실제 어떤 모델인지 매핑)
            # winner가 'A'라면 -> candidates[0]['name']이 승자
            winner_idx = -1
            if res['winner'] == 'A': winner_idx = 0
            elif res['winner'] == 'B': winner_idx = 1
            elif res['winner'] == 'C': winner_idx = 2
            
            real_winner = candidates[winner_idx]['name'] if winner_idx != -1 else "Error"
            
            # 점수 매핑 (A점수 -> A모델 점수)
            score_map = {
                candidates[0]['name']: res['scores'].get('A', 0),
                candidates[1]['name']: res['scores'].get('B', 0),
                candidates[2]['name']: res['scores'].get('C', 0)
            }

            results.append({
                "direction": f"{source_lang}->{target_lang}",
                "category": item['category'],
                "source": item['source'],
                "winner": real_winner, # 실제 모델명 저장
                "score_papago": score_map['Papago'],
                "score_deepl": score_map['DeepL'],
                "score_google": score_map['Google'],
                "reason": res['reason']
            })
            
        except Exception as e:
            print(f"❌ Error: {e}")
            
    return pd.DataFrame(results)

# ==========================================
# 4. 메인 실행
# ==========================================
if __name__ == "__main__":
    # 1. 한국어 -> 영어 평가
    df_k2e = evaluate_translations(KOREAN_TO_ENGLISH_DATA, source_lang="Korean")
    
    # 2. 영어 -> 한국어 평가
    df_e2k = evaluate_translations(ENGLISH_TO_KOREAN_DATA, source_lang="English")
    
    # 3. 결과 합치기
    final_df = pd.concat([df_k2e, df_e2k], ignore_index=True)
    
    if final_df.empty:
        print("\n❌ 평가할 데이터가 없습니다. 상단의 KOREAN_TO_ENGLISH_DATA 등에 번역 결과를 채워주세요.")
    else:
        # 결과 출력
        print("\n" + "="*60)
        print("🏆 최종 스코어보드 (Winner Count)")
        print("="*60)
        print(final_df['winner'].value_counts())
        
        print("\n📊 평균 점수 비교")
        print(f"Papago: {final_df['score_papago'].mean():.2f}")
        print(f"DeepL : {final_df['score_deepl'].mean():.2f}")
        print(f"Google: {final_df['score_google'].mean():.2f}")
        
        # CSV 저장
        final_df.to_csv(OUTPUT_FILE, index=False)
        print(f"\n✅ 상세 리포트 저장 완료: {OUTPUT_FILE}")
```

```python
import json
import os
import random
from datetime import datetime, timedelta
from openai import OpenAI

# OpenAI 클라이언트 설정
# os.environ["OPENAI_API_KEY"] = "sk-..." 
client = OpenAI()

OUTPUT_FILE = "test/summary_eval_dataset_v2.json"

def generate_complex_golden_data():
    # 1. 기준 날짜 설정 (예: 오늘)
    base_date = datetime.now()
    base_date_str = base_date.strftime("%Y-%m-%d")
    
    # 상대 날짜 계산용 힌트 (프롬프트에 주입)
    next_fri = (base_date + timedelta((4 - base_date.weekday()) % 7)).strftime("%Y-%m-%d")
    
    prompt = f"""
    You are a generator for 'Meeting Summary Evaluation Dataset' for a Korean IT company.
    Generate 3 distinct meeting scenarios (Hard, Medium, Easy) in Korean.
    
    [Simulation Setting]
    - Current Date: {base_date_str} (Today)
    - If transcript mentions "next Friday", implies {next_fri}.
    
    [Target Schema per Object]
    1. meeting_id: (Random Number)
    2. meeting_title: String
    3. participants: String (names)
    4. current_date: "{base_date_str}"
    5. transcript: Full dialogue script ("Speaker: Content" format).
       - MUST include relative date references (e.g., "내일까지", "다음 주 수요일에").
    6. ground_truth_json: A JSON object representing the IDEAL OUTPUT. It must contain:
       - summary: 1-line summary.
       - decisions: List of agreed decisions.
       - action_items: List of objects {{ "owner": "Name", "task": "Task", "deadline": "YYYY-MM-DD", "priority": "HIGH/MEDIUM/LOW" }}.
       - risks: List of potential risks mentioned.
    
    [Scenarios]
    1. Hard (Brainstorming & Conflict): 
       - Context: App UI Renewal. Conflicts on design vs dev cost. Slang used.
       - Transcript: 15+ turns.
       - Ground Truth: Must capture the conflict resolution and specific deadlines.
       
    2. Medium (Weekly Report):
       - Context: Bug fixes and Server issues.
       - Transcript: 10-15 turns.
       - Ground Truth: Focus on specific tasks and priorities (HIGH for server issues).
       
    3. Easy (Team Lunch):
       - Context: Choosing a restaurant.
       - Transcript: 8-10 turns.
       - Ground Truth: Simple decision and one action item (Reservation).
       
    Output ONLY the valid JSON list.
    """
    
    print(f">>> GPT-4에게 데이터 생성 요청 중... (기준일: {base_date_str})")
    
    try:
        response = client.chat.completions.create(
            model="gpt-4o",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.7,
            response_format={"type": "json_object"}
        )
        content = response.choices[0].message.content
        
        # JSON 파싱 및 구조화
        if isinstance(content, str):
            data = json.loads(content)
            # { "data": [...] } 형태 대응
            if isinstance(data, dict):
                for key in data:
                    if isinstance(data[key], list):
                        data = data[key]
                        break
        
        # Ragas용 데이터셋으로 변환
        final_dataset = []
        for item in data:
            # Ground Truth JSON을 문자열로 변환 (Ragas는 ground_truth를 문자열로 받음)
            # 나중에 평가할 때 이 구조화된 문자열과 모델의 출력을 비교함
            gt_string = json.dumps(item["ground_truth_json"], ensure_ascii=False, indent=2)
            
            final_dataset.append({
                # [Input for Agent]
                "meeting_id": random.randint(1000, 9999),
                "current_date": base_date_str,
                "meeting_title": item["meeting_title"],
                "participants": item["participants"],
                "transcript": item["transcript"],
                
                # [For Ragas Evaluation]
                "question": "회의록을 분석하여 요약, 결정사항, 액션아이템(기한 포함), 리스크를 추출해줘.",
                "contexts": [item["transcript"]],
                "ground_truth": gt_string  # JSON 형태의 문자열 정답지
            })
            
        # 저장
        with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
            json.dump(final_dataset, f, ensure_ascii=False, indent=2)
            
        print(f"\n✅ 데이터 생성 완료! ({len(final_dataset)}건)")
        print(f"📂 저장 경로: {OUTPUT_FILE}")
        
        # 샘플 확인
        print("\n[Sample Ground Truth Preview]")
        print(final_dataset[0]['ground_truth'][:200] + "...")

    except Exception as e:
        print(f"❌ 생성 실패: {e}")

if __name__ == "__main__":
    generate_complex_golden_data()
```
